```python
import pandas as pd
import numpy as np

# 1. Load cleaned teaching data
df_clean = pd.read_csv('teaching_prices_clean.csv')

# Ensure sorted by date for accurate first-to-last percentage change
df_clean = df_clean.sort_values(by=['item_name', 'date'])

# Compute mean, median, and first-to-last % change per item
summary_stats = []
for item, group in df_clean.groupby('item_name'):
    mean_val = group['unit_price'].mean()
    median_val = group['unit_price'].median()
    first_price = group['unit_price'].iloc[0]
    last_price = group['unit_price'].iloc[-1]
    pct_change = ((last_price - first_price) / first_price) * 100 if first_price != 0 else np.nan

    summary_stats.append({
        'Item': item,
        'Mean': round(mean_val, 2),
        'Median': round(median_val, 2),
        'First_Price': round(first_price, 2),
        'Last_Price': round(last_price, 2),
        'Pct_Change (%)': round(pct_change, 2)
    })

stats_df = pd.DataFrame(summary_stats)
print("--- TEACHING DATA SUMMARY STATISTICS ---")
print(stats_df.to_string(index=False))

# 2. Week 2 Cleaned File: With vs. Without Flagged Value
# Load Week 2 output (update filename if yours differs)
try:
    df_w2 = pd.read_csv('teaching_prices_cleaned_output.csv')
    sample_item = df_w2['item_name'].iloc[0]
    item_data = df_w2[df_w2['item_name'] == sample_item]['unit_price']

    # Simulate/identify a flagged extreme outlier (e.g., max value or > 3 std dev)
    threshold = item_data.mean() + 2 * item_data.std()
    without_flagged = item_data[item_data <= threshold]

    print(f"\n--- FLAGGED VALUE COMPARISON ({sample_item}) ---")
    print(f"With Flagged Value    -> Mean: {item_data.mean():.2f} | Median: {item_data.median():.2f}")
    print(f"Without Flagged Value -> Mean: {without_flagged.mean():.2f} | Median: {without_flagged.median():.2f}")
except FileNotFoundError:
    print("\nUpload your Week 2 CSV file to run the flagged value comparison.")
```
